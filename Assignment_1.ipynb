```python
# Q1. Movie Ticket Counter

age = int(input("Enter your age: "))

if age < 0:
    print("Please enter a valid age.")
elif age < 5:
    print("Your ticket is free.")
elif age <= 17:
    print("Your ticket price is Rs. 100")
elif age <= 59:
    print("Your ticket price is Rs. 200")
else:
    print("Your ticket price is Rs. 120")
```

```python
# Q2. Electricity Bill Calculator

units = int(input("Enter units consumed: "))

if units < 0:
    print("Please enter valid units.")
elif units <= 100:
    bill = units * 3
    print("Total bill: Rs.", bill)
elif units <= 200:
    bill = (100 * 3) + ((units - 100) * 5)
    print("Total bill: Rs.", bill)
else:
    bill = (100 * 3) + (100 * 5) + ((units - 200) * 8)
    print("Total bill: Rs.", bill)
```

```python
# Q3. Multiplication Table for a Kid

number = int(input("Enter a number: "))

for value in range(1, 11):
    result = number * value
    print(number, "x", value, "=", result)
```

```python
# Q4. Savings Piggy Bank

total_savings = 0
days = 0

while total_savings < 1000:
    today_savings = int(input("Enter today's savings: "))

    if today_savings > 0:
        total_savings = total_savings + today_savings
        days = days + 1
    else:
        print("Please enter an amount greater than zero.")

print("Goal reached in", days, "days!")
print("Total saved: Rs.", total_savings)
```

```python
# Q5. ATM PIN Check

correct_pin = 4321
attempts = 0
access_granted = False

while attempts < 3:
    entered_pin = int(input("Enter PIN: "))

    if entered_pin == correct_pin:
        print("Access granted")
        access_granted = True
        break
    else:
        print("Wrong PIN")
        attempts = attempts + 1

if access_granted == False:
    print("Card blocked")
```

```python
# Q6. Even and Odd Counter

number = int(input("Enter N: "))

even_count = 0
odd_count = 0

if number < 1:
    print("Please enter a positive number.")
else:
    for roll_number in range(1, number + 1):
        if roll_number % 2 == 0:
            even_count = even_count + 1
        else:
            odd_count = odd_count + 1

    print("Even numbers:", even_count)
    print("Odd numbers:", odd_count)
```

```python
# Q7. Number Guessing Game

secret_number = 27
guess_count = 0
guess = 0

while guess != secret_number:
    guess = int(input("Guess the number: "))
    guess_count = guess_count + 1

    if guess > secret_number:
        print("Too high")
    elif guess < secret_number:
        print("Too low")
    else:
        print("Correct! You took", guess_count, "guesses.")
```

```python
# Q8. Exam Result with Grade

marks = int(input("Enter marks out of 100: "))

if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
elif marks >= 35:
    print("Grade: D")
else:
    print("Result: Fail")
```

```python
# Q9. Sum of Digits on a Bill Number

bill_number = int(input("Enter bill number: "))

if bill_number < 0:
    print("Please enter a valid bill number.")
else:
    number = bill_number
    digit_sum = 0

    while number > 0:
        digit = number % 10
        digit_sum = digit_sum + digit
        number = number // 10

    print("Sum of digits:", digit_sum)

    if digit_sum > 20:
        print("Congratulations! You win a free gift.")
    else:
        print("Sorry, you do not get a free gift.")
```

```python
# Q10. Staircase Pattern

steps = int(input("Enter number of steps: "))

if steps <= 0:
    print("Please enter a positive number.")
else:
    for row in range(1, steps + 1):
        for column in range(row):
            print("*", end=" ")
        print()
```
