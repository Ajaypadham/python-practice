```python
# Q1. Phone Contact Book

contacts = {}

while True:
    print("\nPhone Contact Book")
    print("1. Add contact")
    print("2. Search contact")
    print("3. Show all contacts")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter contact name: ").title()

        if name in contacts:
            print("Contact already exists")
        else:
            phone_number = input("Enter phone number: ")
            contacts[name] = phone_number
            print("Contact added successfully")

    elif choice == "2":
        name = input("Enter contact name to search: ").title()

        if name in contacts:
            print("Name:", name)
            print("Phone number:", contacts[name])
        else:
            print("Contact not found")

    elif choice == "3":
        if len(contacts) == 0:
            print("No contacts available")
        else:
            print("\nAll Contacts:")

            for name, phone_number in contacts.items():
                print(name, ":", phone_number)

    elif choice == "4":
        print("Contact book closed")
        break

    else:
        print("Invalid choice")
```

```python
# Q3. Student Marks Dictionary

marks = {
    "Aman": 82,
    "Priya": 91,
    "Rahul": 45,
    "Sneha": 67,
    "Kiran": 30
}

passed_students = {}

topper_name = ""
topper_marks = -1

for name, student_marks in marks.items():
    if student_marks >= 40:
        print(name, "-", "Pass")
        passed_students[name] = student_marks
    else:
        print(name, "-", "Fail")

    if student_marks > topper_marks:
        topper_marks = student_marks
        topper_name = name

print("\nTopper:", topper_name)
print("Topper marks:", topper_marks)
print("Passed students:", passed_students)
```

```python
# Q4. Common Friends

riya_friends = {"Aman", "Neha", "Karan", "Sita", "Vikram"}
arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}

mutual_friends = riya_friends.intersection(arjun_friends)
only_riya_friends = riya_friends.difference(arjun_friends)
all_friends = riya_friends.union(arjun_friends)
suggestions_for_riya = arjun_friends.difference(riya_friends)

print("Mutual friends:", sorted(mutual_friends))
print("Friends only Riya has:", sorted(only_riya_friends))
print("All people:", sorted(all_friends))
print("Friend suggestions for Riya:", sorted(suggestions_for_riya))
```

```python
# Q5. Character Counter

text = input("Enter text: ").lower()

character_count = {}

for character in text:
    if character != " ":
        character_count[character] = character_count.get(character, 0) + 1

print(character_count)

if len(character_count) == 0:
    print("No characters were entered.")
else:
    most_frequent_character = ""
    highest_count = 0

    for character, count in character_count.items():
        if count > highest_count:
            highest_count = count
            most_frequent_character = character

    print(
        "Most frequent character:",
        most_frequent_character,
        "(",
        highest_count,
        "times)"
    )
```

```python
# Q6. Restaurant Menu Billing

menu = {
    "dosa": 60,
    "idli": 40,
    "vada": 30,
    "coffee": 25,
    "tea": 15
}

total_bill = 0

print("Restaurant Menu:")

for item, price in menu.items():
    print(item.title(), "- Rs.", price)

while True:
    order = input("Enter an item or type 'bill': ").lower()

    if order == "bill":
        break
    elif order in menu:
        total_bill = total_bill + menu[order]
        print(order.title(), "added - Rs.", menu[order])
    else:
        print("Item not available")

print("\nBill before discount: Rs.", total_bill)

if total_bill > 200:
    discount = total_bill * 10 / 100
    final_bill = total_bill - discount

    print("Discount: Rs.", discount)
    print("Total bill after discount: Rs.", final_bill)
else:
    print("Total bill: Rs.", total_bill)
```

```python
# Q7. City Coordinates (Tuples)

cities = [
    ("Hyderabad", 17.38, 78.48),
    ("Mumbai", 19.07, 72.87),
    ("Chennai", 13.08, 80.27),
    ("Delhi", 28.61, 77.20)
]

highest_latitude = -90
northern_city = ""

for city, latitude, longitude in cities:
    print(
        "City:", city,
        "| Latitude:", latitude,
        "| Longitude:", longitude
    )

    if latitude > highest_latitude:
        highest_latitude = latitude
        northern_city = city

print("\nCity farthest north:", northern_city)
print("Highest latitude:", highest_latitude)

# Trying to change Mumbai's latitude:
# cities[1][1] = 20.00

# Error:
# TypeError: 'tuple' object does not support item assignment
#
# Reason:
# A tuple is immutable, which means its values cannot be changed
# after the tuple has been created.
```

```python
# Q8. Library Book Tracker

books = {
    "Python Basics": 3,
    "Data Science 101": 0,
    "Machine Learning": 2,
    "SQL Guide": 1
}

book_name = input("Enter the book name to borrow: ")

matched_book = ""

for title in books.keys():
    if title.lower() == book_name.lower():
        matched_book = title
        break

if matched_book == "":
    print("Book not found")
elif books[matched_book] == 0:
    print("Currently unavailable")
else:
    books[matched_book] = books[matched_book] - 1
    print("Book issued successfully")

print("Updated books:", books)
```

```python
# Q9. Group Employees by Department

employees = [
    ("Ravi", "IT"),
    ("Anjali", "HR"),
    ("Suresh", "IT"),
    ("Meena", "Finance"),
    ("Arun", "HR"),
    ("Divya", "IT")
]

departments = {}

for name, department in employees:
    if department not in departments:
        departments[department] = []

    departments[department].append(name)

print("Employees grouped by department:")
print(departments)

print("\nNumber of employees in each department:")

for department, employee_names in departments.items():
    print(department, ":", len(employee_names))
```

```python
# Q10. Quiz Answer Checker

answer_key = ("B", "C", "A", "D", "B")

wrong_questions = set()
score = 0

for index in range(len(answer_key)):
    student_answer = input(
        "Enter answer for question " + str(index + 1) + ": "
    ).upper()

    if student_answer == answer_key[index]:
        print("Question", index + 1, "is correct")
        score = score + 1
    else:
        print("Question", index + 1, "is wrong")
        wrong_questions.add(index + 1)

print("\nWrong question numbers:", wrong_questions)
print("Final score:", score, "out of 5")

if score == 5:
    print("Remark: Excellent")
elif score >= 3:
    print("Remark: Good")
else:
    print("Remark: Needs improvement")
```
