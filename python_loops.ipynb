# The range() Function

The range() function generates a sequence of numbers.

It is commonly used with a for loop.

Syntax:

range(start, stop, step)

```python
# Basic range()

numbers = range(5)

print(numbers)
print(list(numbers))
```

# range(stop)
When only one number is provided, Python starts from `0` and stops before the given number.

```python
print(list(range(5)))
print(list(range(10)))
```

# range(start, stop)

The first number specifies where the sequence starts.

The second number specifies where it stops. The stopping value is not included.

```python
print(list(range(1, 6)))
print(list(range(5, 11)))
```

# range(start, stop, step)

The third value determines how much Python increases the number each time.

```python
print(list(range(0, 11, 2)))
print(list(range(1, 11, 2)))
```

```python
# Counting backward

Use a negative step to generate numbers in reverse:
```

```python
print(list(range(10, 0, -1)))
```

# Using range() with a for loop

```python
for number in range(1, 6):
    print(number)
```

# For Loops

A for loop repeats a block of code for every item in a sequence.

A sequence can be a range, string, list, tuple, or dictionary.

```python
# Basic for loop with range()
for number in range(1, 6):
    print(number)
```

# Understanding indentation

The indented lines belong to the loop:

```python
for number in range(1, 4):
    print("Current number:", number)
    print("Inside the loop")

print("Loop completed")
```

```python
# Even numbers
for number in range(2, 11, 2):
    print(number)
```

```python
# Loop through a string

course = "Python"

for letter in course:
    print(letter)
```

```python
# Loop through a list

technologies = ["Python", "SQL", "Databricks", "Azure"]

for technology in technologies:
    print(technology)
```

# While Loops

A while loop repeatedly executes code as long as its condition is True.

The variable used in the condition must normally be updated inside the loop. Otherwise, the loop may continue forever.

```python
# Basic while loop

number = 1

while number <= 5:
    print(number)
    number = number + 1
```

```python
# Countdown using a while loop

number = 5

while number >= 1:
    print(number)
    number -= 1

print("Countdown completed!")
```

# Break and Continue

The `break` statement immediately stops a loop.

The `continue` statement skips the current iteration and continues with the next iteration.

```python
# Break statement

for number in range(1, 11):
    if number == 6:
        break

    print(number)
```

```python
# Continue statement

for number in range(1, 11):
    if number == 6:
        continue

    print(number)
```
