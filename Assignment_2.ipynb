```python
# Q1. Class Marks Report

marks = [78, 45, 92, 33, 67, 88, 51, 29]

highest_marks = max(marks)
lowest_marks = min(marks)
average_marks = sum(marks) / len(marks)

passed_students = 0

for mark in marks:
    if mark >= 35:
        passed_students = passed_students + 1

print("Highest marks:", highest_marks)
print("Lowest marks:", lowest_marks)
print("Class average:", average_marks)
print("Number of students who passed:", passed_students)
```

```python
# Q2. Grocery Shopping List

shopping_list = []

while True:
    item = input("Enter item: ").lower()

    if item == "done":
        break
    elif item in shopping_list:
        print("Already added")
    else:
        shopping_list.append(item)

print("\nShopping List:")

for index in range(len(shopping_list)):
    print(index + 1, ".", shopping_list[index])

print("Total items:", len(shopping_list))
```

```python
# Q3. Username Validator

username = input("Enter username: ")

if len(username) < 5 or len(username) > 12:
    print("Invalid: Username must contain between 5 and 12 characters")
elif " " in username:
    print("Invalid: Username cannot contain spaces")
elif username[0].isdigit():
    print("Invalid: Username cannot start with a digit")
else:
    print("Valid username")
```

```python
# Q4. Vowel Counter for a Tweet

sentence = input("Enter a sentence: ").lower()

vowel_count = 0
consonant_count = 0
space_count = 0

for character in sentence:
    if character in "aeiou":
        vowel_count = vowel_count + 1
    elif character == " ":
        space_count = space_count + 1
    elif character.isalpha():
        consonant_count = consonant_count + 1

print("Number of vowels:", vowel_count)
print("Number of consonants:", consonant_count)
print("Number of spaces:", space_count)
```

```python
# Q5. Palindrome Name Checker

name = input("Enter your name: ").lower()

reversed_name = name[::-1]

if name == reversed_name:
    print("Bonus! Your name is a palindrome.")
else:
    print("Your name is not a palindrome.")
```

```python
# Q6. Temperature Tracker

temps = [32, 35, 38, 41, 36, 29, 40]
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

hottest_temperature = temps[0]
hottest_day = days[0]
fahrenheit_temps = []

print("Heat alert days:")

for index in range(len(temps)):
    if temps[index] > 37:
        print(days[index], "-", temps[index], "°C")

    if temps[index] > hottest_temperature:
        hottest_temperature = temps[index]
        hottest_day = days[index]

    fahrenheit = temps[index] * 9 / 5 + 32
    fahrenheit_temps.append(fahrenheit)

print("\nHottest day:", hottest_day)
print("Hottest temperature:", hottest_temperature, "°C")
print("Temperatures in Fahrenheit:", fahrenheit_temps)
```

```python
# Q7. Word Frequency in a Sentence

sentence = input("Enter a sentence: ")

words = sentence.split()

if len(words) == 0:
    print("Please enter a sentence.")
else:
    longest_word = max(words, key=len)
    shortest_word = min(words, key=len)

    capital_count = 0

    for word in words:
        if word[0].isupper():
            capital_count = capital_count + 1

    print("Longest word:", longest_word)
    print("Shortest word:", shortest_word)
    print("Words starting with capital letter:", capital_count)
```

```python
# Q8. Remove Negative Balances

balances = [1500, -200, 3000, -50, 0, 750, -1000]

positive_accounts = []
negative_accounts = []
total_amount_owed = 0

for balance in balances:
    if balance > 0:
        positive_accounts.append(balance)
    elif balance < 0:
        negative_accounts.append(balance)
        total_amount_owed = total_amount_owed + (-balance)

print("Positive accounts:", positive_accounts)
print("Negative accounts:", negative_accounts)
print("Total amount owed:", total_amount_owed)
```

```python
# Q9. Password Strength Checker

password = input("Enter a password: ")

has_uppercase = False
has_lowercase = False
has_digit = False
has_minimum_length = False

for character in password:
    if character.isupper():
        has_uppercase = True
    elif character.islower():
        has_lowercase = True
    elif character.isdigit():
        has_digit = True

if len(password) >= 8:
    has_minimum_length = True

conditions_met = 0

if has_uppercase:
    conditions_met = conditions_met + 1

if has_lowercase:
    conditions_met = conditions_met + 1

if has_digit:
    conditions_met = conditions_met + 1

if has_minimum_length:
    conditions_met = conditions_met + 1

if conditions_met == 4:
    print("Password strength: Strong")
elif conditions_met == 2 or conditions_met == 3:
    print("Password strength: Medium")
else:
    print("Password strength: Weak")
```
