## 4. Type Casting

Type casting means converting a value from one data type to another.

Common conversion functions:

int() converts a value into an integer.
float()converts a value into a decimal number.
str() converts a value into text.
bool() converts a value into a Boolean.

```python
price = "100"

print(price)
print(type(price))

converted_price = int(price)

print(converted_price)
print(type(converted_price))

# converting to float.
# float converts number into a decimal number.

number = 25
decimal_number = float(number)

print(number, type(number))
print(decimal_number, type(decimal_number))

# Converting to a string

age = 25
converted_age = str(age)

print(age, type(age))
print(converted_age, type(converted_age))
```

# Operators in Python

Operators are symbols that perform operations on values and variables.

The main operator categories are:

1. Arithmetic operators
2. Comparison operators
3. Assignment operators
4. Logical operators
5. Membership operators

# Arithmetic Operators

Arithmetic operators perform mathematical calculations.

| Operator | Operation      | Example   |
| -------- | -------------- | --------- |
| `+`      | Addition       | `10 + 5`  |
| `-`      | Subtraction    | `10 - 5`  |
| `*`      | Multiplication | `10 * 5`  |
| `/`      | Division       | `10 / 5`  |
| `//`     | Floor division | `10 // 3` |
| `%`      | Remainder      | `10 % 3`  |
| `**`     | Exponent       | `2 ** 3`  |


```python
first_number = 10
second_number = 3

print("Addition:", first_number + second_number)
print("Subtraction:", first_number - second_number)
print("Multiplication:", first_number * second_number)
print("Division:", first_number / second_number)
print("Floor division:", first_number // second_number)
print("Remainder:", first_number % second_number)
print("Exponent:", first_number ** second_number)
```

# Comparison operators

Comparison operators compare two values. Their result is always True or False.

| Operator | Meaning                  |
| -------- | ------------------------ |
| `==`     | Equal to                 |
| `!=`     | Not equal to             |
| `>`      | Greater than             |
| `<`      | Less than                |
| `>=`     | Greater than or equal to |
| `<=`     | Less than or equal to    |


```python
first_number = 20
second_number = 10

print(first_number == second_number)
print(first_number != second_number)
print(first_number > second_number)
print(first_number < second_number)
print(first_number >= 20)
print(second_number <= 10)
```

```python
# Assignment operators
number = 10
print(number)

number += 5
print(number)

number -= 3
print(number)

number *= 2
print(number)

number /= 4
print(number)

```

# Logical operators

Logical operators combine conditions.

| Operator | Meaning                             |
| -------- | ----------------------------------- |
| `and`    | Both conditions must be true        |
| `or`     | At least one condition must be true |
| `not`    | Reverses the result                 |


```python
age = 25
has_id = True

print(age >= 18 and has_id == True)
print(age < 18 or has_id == True)
print(not has_id)
```

# Membership operators

Membership operators check whether a value exists inside another value.

```python
course = "Python Programming"

print("Python" in course)
print("SQL" in course)
print("Java" not in course)
```

# F-Strings

An f-string is used to insert variables and expressions directly inside a string.

Place the letter f before the quotation marks and write variables inside curly braces {}.

```python
name = "Ajay"
job = "Release Engineer"

print(f"My name is {name}.")
print(f"I work as a {job}.")
```

```python
# multiple variables

name = "Ajay"
city = "Charlotte"
experience = 5

print(f"My name is {name}, I live in {city}, and I have {experience} years of experience.")
```

```python
# f-string with user input

name = input("Enter your name: ")
age = int(input("Enter your age: "))
city = input("Enter your city: ")

print(f"My name is {name}.")
print(f"I am {age} years old.")
print(f"I live in {city}.")
print(f"Next year, I will be {age + 1} years old.")
```

# The .format method
The .format() method is another way to insert variable values into a string. It was commonly used before f-strings.

```python
name = "Ajay"
job_title = "Release Engineer"

print("My name is {}.".format(name))
print("I work as a {}.".format(job_title))
```

```python
# multiple variables
name = "Ajay"
city = "Charlotte"
experience = 5

print("My name is {}, I live in {}, and I have {} years of experience."
      .format(name, city, experience))

# Numbered Placeholders

#Numbers inside the placeholders control the position of each value.
#Python begins counting positions from zero.

name = "Ajay"
course = "Python"

print("{0} is learning {1}.".format(name, course))
print("{1} is being learned by {0}.".format(name, course))

# named placeholder

print(
    "{employee} works as a {job}."
    .format(employee="Ajay", job="Release Engineer")
)

# F-string versus .format()

name = "Ajay"
experience = 5

# Using an f-string
print(f"{name} has {experience} years of experience.")

# Using .format()
print("{} has {} years of experience.".format(name, experience))
```

# Branching: if, elif, and else

Branching allows Python to choose which block of code to execute.

"if" checks the first condition.
"elif" checks another condition when the previous condition is false.
"else" executes when none of the previous conditions are true.

```python
# simple if statement

age = 25

if age >= 18:
    print("You are an adult")
```

```python
# if and else statement

age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible")
else:
    print("You are not eligible")
```

```python
# using if, elif, else

# Multiple Conditions:
#Use "elif" when the program needs to check more than two possibilities.

score = int(input("Enter your score: "))

if score >= 90:
    print("Grade A")
elif score >= 75:
    print("Grade B")
elif score >= 50:
    print("Grade C")
else:
    print("Fail")
```

```python
# Multiple conditions with and

age = int(input("Enter your age: "))
has_id = input("Do you have an ID? Enter yes or no: ")

if age >= 18 and has_id.lower() == "yes":
    print("You are allowed to enter")
else:
    print("You are not allowed to enter")
```

```python
# Nested if statements
#A nested if means an if statement inside another if statement.

age = int(input("Enter your age: "))
has_id = input("Do you have an ID? Enter yes or no: ")

if age >= 18:
    if has_id.lower() == "yes":
        print("You are allowed to enter")
    else:
        print("You need an ID")
else:
    print("You must be at least 18")
```
